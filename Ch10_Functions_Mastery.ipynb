## 10.1 — Function Basics

```python
def greet(name, greeting="Hello"):
    """Greet someone with an optional greeting."""
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", "Hi"))

# Keyword arguments (order doesn't matter)
print(greet(greeting="Hey", name="Charlie"))
```

## 10.2 — `*args` and `**kwargs`
Accept any number of positional or keyword arguments.

```python
def flexible(*args, **kwargs):
    print(f"args = {args}")      # Tuple of positional args
    print(f"kwargs = {kwargs}")  # Dict of keyword args

flexible(1, 2, 3, name="Alice", age=30)

# Practical example: a function that wraps print
def debug_print(*values, label="DEBUG", **print_kwargs):
    print(f"[{label}]", *values, **print_kwargs)

debug_print("x =", 42, "y =", 99)
debug_print("Loading", label="INFO", end="...\n")
```

## 10.3 — 🔥 Unpacking Arguments
The `*` and `**` operators work in reverse too — to UNPACK args.

```python
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add(*nums))  # Unpacks list → add(1, 2, 3)

config = {"a": 10, "b": 20, "c": 30}
print(add(**config))  # Unpacks dict → add(a=10, b=20, c=30)
```

## 10.4 — Lambda Functions
Anonymous, single-expression functions.

```python
# Regular function
def square(x):
    return x ** 2

# Same thing as a lambda
square_l = lambda x: x ** 2

print(square(5), square_l(5))  # Both: 25

# 🔥 Lambdas shine when used with sorted(), map(), filter()
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]

# Sort by score
by_score = sorted(students, key=lambda s: s[1], reverse=True)
print("By score:", by_score)

# Filter: keep only scores above 80
above_80 = list(filter(lambda s: s[1] > 80, students))
print("Above 80:", above_80)

# Map: extract just the names
names = list(map(lambda s: s[0], students))
print("Names:", names)
```

## 10.5 — Scope: The LEGB Rule
Python looks for variables in this order:
1. **L**ocal — inside the current function
2. **E**nclosing — in the enclosing function (closures)
3. **G**lobal — at the module level
4. **B**uilt-in — Python's built-in names

```python
x = "global"

def outer():
    x = "enclosing"
    
    def inner():
        x = "local"
        print(f"inner sees: {x}")
    
    inner()
    print(f"outer sees: {x}")

outer()
print(f"module sees: {x}")
```

## 10.6 — Functions as First-Class Objects

```python
# Functions can be assigned to variables
def shout(text):
    return text.upper() + "!"

yell = shout  # yell is now the same function
print(yell("hello"))

# Functions can be passed as arguments
def apply(func, value):
    return func(value)

print(apply(len, "Python"))     # 6
print(apply(sorted, [3, 1, 2]))  # [1, 2, 3]

# Functions can return other functions (function factory)
def multiplier(n):
    def multiply(x):
        return x * n
    return multiply

double = multiplier(2)
triple = multiplier(3)
print(double(5))  # 10
print(triple(5))  # 15
```

## 10.7 — 🔥 Trick: Keyword-Only and Positional-Only Args (Python 3.8+)

```python
# Everything after * is keyword-only
def api_call(url, *, method="GET", timeout=30):
    print(f"{method} {url} (timeout={timeout}s)")

api_call("https://example.com", method="POST")
# api_call("https://example.com", "POST")  # ❌ TypeError!

# Everything before / is positional-only
def greet(name, /, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Alice"))
# greet(name="Alice")  # ❌ TypeError!
```
