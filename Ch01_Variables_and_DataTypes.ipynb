## 1.1 — Your First Variables
Python is **dynamically typed** — you don't declare types, Python figures it out.

```python
# Variables — no 'int x = 5;' needed!
name = "Alice"
age = 25
height = 5.7
is_student = True
nothing = None

print(name, age, height, is_student, nothing)
```

## 1.2 — Checking Types with `type()`

```python
print(type(name))       # <class 'str'>
print(type(age))        # <class 'int'>
print(type(height))     # <class 'float'>
print(type(is_student)) # <class 'bool'>
print(type(nothing))    # <class 'NoneType'>
```

## 1.3 — `print()` Power Tricks

### Trick 1: `sep` and `end` parameters

```python
# Custom separator
print("2025", "10", "07", sep="-")   # 2025-10-07

# Custom ending (default is '\n')
print("Loading", end="...")
print("Done!")  # prints on the same line
```

### Trick 2: f-strings (formatted string literals) — the BEST way to print

```python
name = "Bob"
score = 95.678

# f-strings — embed expressions directly!
print(f"Hello, {name}!")
print(f"Score: {score:.2f}")          # 2 decimal places
print(f"{'Centered':^20}")            # Center in 20 chars
print(f"{'Left':<20}|")              # Left align
print(f"{'Right':>20}|")             # Right align
print(f"Binary of 42: {42:b}")        # Binary
print(f"Hex of 255: {255:#x}")        # Hex
print(f"Big number: {1000000:,}")     # Thousands separator
```

## 1.4 — Type Conversion

```python
# str → int → float
x = "42"
print(int(x) + 8)       # 50
print(float(x) + 0.5)   # 42.5

# float → int (truncates, does NOT round)
print(int(3.9))   # 3  (not 4!)
print(round(3.9)) # 4  (use round() to round)

# Anything → str
print(str(123))          # "123"
print(str(True))         # "True"

# Truthiness: what converts to True/False?
print(bool(0), bool(1), bool(-1))        # False True True
print(bool(""), bool("hello"))           # False True
print(bool([]), bool([1, 2]))            # False True
print(bool(None))                        # False
```

## 1.5 — Multiple Assignment Tricks

```python
# Assign multiple variables in one line
a, b, c = 1, 2, 3
print(a, b, c)

# Same value to multiple variables
x = y = z = 0
print(x, y, z)

# Swap without a temp variable!
a, b = 10, 20
a, b = b, a
print(f"a={a}, b={b}")  # a=20, b=10
```
