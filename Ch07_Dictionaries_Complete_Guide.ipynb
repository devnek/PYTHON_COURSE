## 7.1 — Creating Dictionaries

```python
# Various ways to create dicts
d1 = {"name": "Alice", "age": 30}
d2 = dict(name="Alice", age=30)          # Keyword syntax
d3 = dict([("name", "Alice"), ("age", 30)])  # From pairs
d4 = dict.fromkeys(["a", "b", "c"], 0)   # All values = 0

print(d4)  # {'a': 0, 'b': 0, 'c': 0}
```

## 7.2 — Safe Access Patterns

```python
user = {"name": "Alice", "age": 30}

# ❌ Dangerous — raises KeyError if key missing
# print(user["email"])

# ✅ Safe — returns default if missing
print(user.get("email", "N/A"))  # N/A

# ✅ setdefault — get value, OR set it if missing
user.setdefault("role", "member")
print(user)  # {'name': 'Alice', 'age': 30, 'role': 'member'}
# If key already exists, setdefault does NOT overwrite
user.setdefault("role", "admin")
print(user["role"])  # Still 'member'
```

## 7.3 — Iterating Dicts

```python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}

# Iterate keys (default)
for name in scores:
    print(name)

# Iterate values
for score in scores.values():
    print(score)

# Iterate key-value pairs (most useful!)
for name, score in scores.items():
    print(f"{name}: {score}")
```

## 7.4 — Dictionary Comprehensions

```python
# Basic
squares = {x: x**2 for x in range(6)}
print(squares)  # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# With condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(even_squares)

# Invert a dictionary (swap keys and values)
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(inverted)  # {1: 'a', 2: 'b', 3: 'c'}

# From two lists
keys = ["name", "age", "city"]
values = ["Alice", 30, "NYC"]
combined = dict(zip(keys, values))
print(combined)
```

## 7.5 — 🔥 Pro Dict Tricks

```python
# Trick 1: Merge dicts (Python 3.9+)
d1 = {"a": 1, "b": 2}
d2 = {"b": 3, "c": 4}
merged = d1 | d2          # d2 values win on conflicts
print("Merged:", merged)   # {'a': 1, 'b': 3, 'c': 4}

# For Python < 3.9:
merged_old = {**d1, **d2}
print("Merged (old):", merged_old)

# Trick 2: Count occurrences using dict
text = "mississippi"
freq = {}
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1
print("Frequency:", freq)

# Trick 3: Group items by a key
from collections import defaultdict
words = ["apple", "avocado", "banana", "blueberry", "cherry", "cranberry"]
grouped = defaultdict(list)
for word in words:
    grouped[word[0]].append(word)
print("Grouped:", dict(grouped))

# Trick 4: Get key with max/min value
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}
best = max(scores, key=scores.get)
worst = min(scores, key=scores.get)
print(f"Best: {best}, Worst: {worst}")

# Trick 5: Pretty print dicts
import json
data = {"users": [{"name": "Alice", "scores": [85, 92]}, {"name": "Bob", "scores": [78, 88]}]}
print(json.dumps(data, indent=2))
```
