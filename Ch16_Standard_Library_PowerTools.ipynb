## 16.1 — `collections` Module

```python
from collections import Counter, defaultdict, deque, OrderedDict

# ─── Counter ───
words = "the quick brown fox jumps over the lazy dog the fox".split()
word_count = Counter(words)
print("Word counts:", word_count)
print("Top 3:", word_count.most_common(3))

# Counter arithmetic
c1 = Counter("aabbc")
c2 = Counter("abbcc")
print(f"c1 + c2 = {c1 + c2}")   # Add counts
print(f"c1 - c2 = {c1 - c2}")   # Subtract (keeps positive only)
print(f"c1 & c2 = {c1 & c2}")   # Min of each (intersection)
print(f"c1 | c2 = {c1 | c2}")   # Max of each (union)
```

```python
# ─── defaultdict ───
# Group words by first letter
grouped = defaultdict(list)
for word in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    grouped[word[0]].append(word)
print("Grouped:", dict(grouped))

# Count with defaultdict(int)
freq = defaultdict(int)
for ch in "mississippi":
    freq[ch] += 1
print("Freq:", dict(freq))
```

```python
# ─── deque (double-ended queue) ───
# O(1) append/pop from BOTH ends (lists are O(n) for pop(0))
d = deque([1, 2, 3])
d.appendleft(0)     # Add to left
d.append(4)         # Add to right
print(f"deque: {d}")

d.popleft()          # Remove from left — O(1)!
print(f"After popleft: {d}")

# Rotation
d = deque([1, 2, 3, 4, 5])
d.rotate(2)          # Rotate right by 2
print(f"Rotated right: {d}")
d.rotate(-2)         # Rotate left by 2
print(f"Rotated back: {d}")

# 🔥 Bounded deque — auto-removes oldest when full (great for "last N items")
recent = deque(maxlen=3)
for i in range(6):
    recent.append(i)
    print(f"  Added {i}: {list(recent)}")
```

## 16.2 — `itertools` Module

```python
import itertools

# ─── Combinations and Permutations ───
items = ['A', 'B', 'C', 'D']
print("Combinations (2):", list(itertools.combinations(items, 2)))
print("Permutations (2):", list(itertools.permutations(items, 2)))

# Combinations with replacement
print("Combos w/ replace:", list(itertools.combinations_with_replacement("AB", 2)))

# Cartesian product
print("Product:", list(itertools.product("AB", "12")))
# [('A','1'), ('A','2'), ('B','1'), ('B','2')]
```

```python
# ─── Chaining and Grouping ───

# chain — concatenate iterables
a = [1, 2, 3]
b = [4, 5, 6]
c = [7, 8, 9]
print("Chained:", list(itertools.chain(a, b, c)))

# chain.from_iterable — flatten one level
nested = [[1, 2], [3, 4], [5, 6]]
print("Flattened:", list(itertools.chain.from_iterable(nested)))

# groupby — group consecutive elements
data = sorted(["apple", "avocado", "banana", "blueberry", "cherry"], key=lambda x: x[0])
for letter, group in itertools.groupby(data, key=lambda x: x[0]):
    print(f"  {letter}: {list(group)}")
```

```python
# ─── Infinite iterators ───

# count — infinite counter
for i in itertools.islice(itertools.count(10, 2), 5):
    print(i, end=" ")  # 10 12 14 16 18
print()

# cycle — infinite cycle through an iterable
colors = itertools.cycle(["red", "green", "blue"])
for _, color in zip(range(7), colors):
    print(color, end=" ")
print()

# repeat
print(list(itertools.repeat("hello", 3)))

# accumulate — running totals
nums = [1, 2, 3, 4, 5]
print("Running sum:", list(itertools.accumulate(nums)))
print("Running max:", list(itertools.accumulate(nums, max)))
```

## 16.3 — `functools` Module

```python
from functools import reduce, partial

# ─── reduce — fold a sequence into a single value ───
nums = [1, 2, 3, 4, 5]
product = reduce(lambda a, b: a * b, nums)
print(f"Product of {nums} = {product}")  # 120

# Find max using reduce (just for illustration; use max() normally)
biggest = reduce(lambda a, b: a if a > b else b, nums)
print(f"Biggest: {biggest}")

# ─── partial — freeze some arguments ───
def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)

print(f"square(5) = {square(5)}")  # 25
print(f"cube(3) = {cube(3)}")      # 27

# 🔥 Useful with sorting
from functools import cmp_to_key

def compare(a, b):
    """Custom comparison: sort by length, then alphabetically."""
    if len(a) != len(b):
        return len(a) - len(b)
    return (a > b) - (a < b)

words = ["banana", "pie", "a", "cherry", "hi"]
print(sorted(words, key=cmp_to_key(compare)))
```

## 16.4 — `datetime` Module

```python
from datetime import datetime, date, timedelta

# Current date and time
now = datetime.now()
print(f"Now: {now}")
print(f"Date: {now.date()}")
print(f"Time: {now.time()}")

# Formatting
print(f"Formatted: {now.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Human: {now.strftime('%B %d, %Y at %I:%M %p')}")

# Parsing strings to dates
birthday = datetime.strptime("1995-06-15", "%Y-%m-%d")
print(f"Birthday: {birthday.date()}")

# Date arithmetic with timedelta
tomorrow = date.today() + timedelta(days=1)
next_week = date.today() + timedelta(weeks=1)
print(f"Tomorrow: {tomorrow}")
print(f"Next week: {next_week}")

# Difference between dates
age_days = (date.today() - birthday.date()).days
print(f"Days since birthday: {age_days:,}")
```

## 16.5 — `re` (Regular Expressions)

```python
import re

text = "Contact us at support@example.com or sales@company.org. Call 123-456-7890."

# Find all email addresses
emails = re.findall(r'[\w.+-]+@[\w-]+\.[\w.]+', text)
print(f"Emails: {emails}")

# Find phone numbers
phones = re.findall(r'\d{3}-\d{3}-\d{4}', text)
print(f"Phones: {phones}")

# Search — returns first match
match = re.search(r'(\w+)@(\w+)', text)
if match:
    print(f"First email: user={match.group(1)}, domain={match.group(2)}")

# Sub — replace
censored = re.sub(r'\d', '*', text)
print(f"Censored: {censored}")

# Split on multiple delimiters
data = "apple;banana,cherry:date"
items = re.split(r'[;,:]', data)
print(f"Split: {items}")
```

## 16.6 — 🔥 Bonus: `string` Module Constants

```python
import string

print(f"ASCII lowercase: {string.ascii_lowercase}")
print(f"ASCII uppercase: {string.ascii_uppercase}")
print(f"Digits: {string.digits}")
print(f"Punctuation: {string.punctuation}")

# 🔥 Generate a random password using only stdlib!
import random
import secrets  # cryptographically secure

alphabet = string.ascii_letters + string.digits + "!@#$%"
# Use secrets for real passwords (not random)
password = ''.join(secrets.choice(alphabet) for _ in range(16))
print(f"Random password: {password}")
```
