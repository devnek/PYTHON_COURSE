## 13.1 — Text File Operations

```python
# Writing
with open("demo.txt", "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")
    f.writelines(["Line 3\n", "Line 4\n"])

# Reading — multiple ways
with open("demo.txt", "r") as f:
    # Method 1: Read entire file
    content = f.read()
    print("--- Full content ---")
    print(content)

with open("demo.txt", "r") as f:
    # Method 2: Read all lines into a list
    lines = f.readlines()
    print("Lines:", lines)

with open("demo.txt", "r") as f:
    # Method 3: Iterate line by line (memory efficient!)
    print("--- Line by line ---")
    for line in f:
        print(line.strip())

# Appending
with open("demo.txt", "a") as f:
    f.write("Line 5 (appended)\n")
```

## 13.2 — `pathlib` — Modern Path Handling

```python
from pathlib import Path

# Create Path objects
p = Path("demo.txt")
print(f"Exists: {p.exists()}")
print(f"Is file: {p.is_file()}")
print(f"Name: {p.name}")
print(f"Stem: {p.stem}")        # Filename without extension
print(f"Suffix: {p.suffix}")    # Extension
print(f"Parent: {p.parent}")
print(f"Absolute: {p.resolve()}")

# Read/write with pathlib (one-liners!)
Path("pathlib_demo.txt").write_text("Hello from pathlib!")
content = Path("pathlib_demo.txt").read_text()
print(content)

# List files in a directory
print("\nPython files here:")
for f in Path(".").glob("*.py"):
    print(f"  {f}")

# Recursive glob
print("\nAll .txt files (recursive):")
for f in Path(".").rglob("*.txt"):
    print(f"  {f}")
```

## 13.3 — JSON (Native Module)

```python
import json

# Python dict → JSON string
data = {
    "name": "Alice",
    "age": 30,
    "scores": [85, 92, 78],
    "active": True,
    "address": None
}

# Serialize to string
json_str = json.dumps(data, indent=2)
print(json_str)

# Parse JSON string back to Python
parsed = json.loads(json_str)
print(f"Name: {parsed['name']}, Scores: {parsed['scores']}")

# Write to file
with open("data.json", "w") as f:
    json.dump(data, f, indent=2)

# Read from file
with open("data.json", "r") as f:
    loaded = json.load(f)
    print(f"Loaded: {loaded}")
```

## 13.4 — CSV (Native Module)

```python
import csv

# Writing CSV
students = [
    ["Name", "Score", "Grade"],
    ["Alice", 85, "B"],
    ["Bob", 92, "A"],
    ["Charlie", 78, "C"],
]

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(students)

# Reading CSV
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    header = next(reader)  # Skip header
    print(f"Header: {header}")
    for row in reader:
        print(f"  {row[0]}: {row[1]} ({row[2]})")

# 🔥 DictReader / DictWriter — use column names!
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"  {row['Name']} scored {row['Score']}")
```

## 13.5 — `os` and `shutil` — File System Operations

```python
import os
import shutil

# Current directory
print(f"CWD: {os.getcwd()}")

# Create directories
os.makedirs("test_dir/sub_dir", exist_ok=True)

# List directory contents
print(f"Contents: {os.listdir('test_dir')}")

# File info
stat = os.stat("demo.txt")
print(f"Size: {stat.st_size} bytes")

# Copy file
shutil.copy2("demo.txt", "test_dir/demo_copy.txt")

# Clean up
shutil.rmtree("test_dir")
print("Cleaned up test_dir")
```
