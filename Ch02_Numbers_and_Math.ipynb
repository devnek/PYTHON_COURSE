## 2.1 — Arithmetic Operators

```python
a, b = 17, 5

print(f"a + b  = {a + b}")     # Addition: 22
print(f"a - b  = {a - b}")     # Subtraction: 12
print(f"a * b  = {a * b}")     # Multiplication: 85
print(f"a / b  = {a / b}")     # True division: 3.4 (always float)
print(f"a // b = {a // b}")    # Floor division: 3 (integer result)
print(f"a % b  = {a % b}")     # Modulo (remainder): 2
print(f"a ** b = {a ** b}")    # Power: 1419857
```

## 2.2 — Trick: Using divmod()
Get quotient AND remainder in one call.

```python
minutes = 137
hours, mins = divmod(minutes, 60)
print(f"{minutes} minutes = {hours}h {mins}m")  # 2h 17m
```

## 2.3 — The `math` Module (Built-in!)

```python
import math

print(f"π = {math.pi}")
print(f"e = {math.e}")
print(f"√144 = {math.sqrt(144)}")
print(f"log₂(1024) = {math.log2(1024)}")
print(f"ceil(4.2) = {math.ceil(4.2)}")    # Round UP → 5
print(f"floor(4.8) = {math.floor(4.8)}")  # Round DOWN → 4
print(f"factorial(6) = {math.factorial(6)}")  # 720
print(f"gcd(48, 18) = {math.gcd(48, 18)}")   # 6
```

## 2.4 — Number Tricks

```python
# Trick 1: Underscores in large numbers (readability)
population = 1_400_000_000
print(population)  # 1400000000 — Python ignores the underscores

# Trick 2: Infinity
inf = float('inf')
print(inf > 999999999999)   # True
print(-float('inf'))         # Negative infinity

# Trick 3: Check if a float is an integer
x = 10.0
print(x.is_integer())  # True
y = 10.5
print(y.is_integer())  # False

# Trick 4: Absolute value
print(abs(-42))  # 42

# Trick 5: Round with precision
print(round(3.14159, 2))  # 3.14

# Trick 6: Sum, min, max on sequences
nums = [4, 7, 1, 9, 3]
print(f"sum={sum(nums)}, min={min(nums)}, max={max(nums)}")
```

## 2.5 — Chained Comparisons

```python
# Python supports mathematical chaining!
x = 5
print(1 < x < 10)       # True — no need for (1 < x) and (x < 10)
print(1 < x < 3)        # False
print(0 <= x <= 5 <= 10) # True — chain as many as you like!
```
