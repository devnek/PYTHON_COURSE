## 6.1 — Matrix Multiplication

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Three equivalent ways
print(f"np.dot:\n{np.dot(A, B)}")
print(f"\nnp.matmul:\n{np.matmul(A, B)}")
print(f"\n@ operator:\n{A @ B}")

# ⚠️ * is element-wise, @ is matrix multiplication!
print(f"\nA * B (element-wise):\n{A * B}")
print(f"\nA @ B (matrix mult):\n{A @ B}")

# Vector dot product
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
print(f"\nDot product: {np.dot(v1, v2)}")  # 32
```

## 6.2 — `np.linalg` Functions

```python
A = np.array([[2, 1], [5, 3]])

# Determinant
print(f"det(A) = {np.linalg.det(A):.4f}")

# Inverse
A_inv = np.linalg.inv(A)
print(f"\nA_inv =\n{A_inv}")
print(f"A @ A_inv =\n{np.round(A @ A_inv, 10)}")  # Should be identity

# Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"\nEigenvalues: {eigenvalues}")
print(f"Eigenvectors:\n{eigenvectors}")

# Matrix norm
print(f"\nFrobenius norm: {np.linalg.norm(A):.4f}")
print(f"L2 norm (vector): {np.linalg.norm([3, 4]):.4f}")  # 5.0

# Rank
M = np.array([[1, 2, 3], [2, 4, 6], [1, 1, 1]])
print(f"\nRank: {np.linalg.matrix_rank(M)}")  # 2 (row 2 = 2 * row 1)
```

## 6.3 — Solving Systems of Linear Equations

```python
# Solve:  2x + y = 5
#         3x + 2y = 8

A = np.array([[2, 1], [3, 2]])  # Coefficients
b = np.array([5, 8])             # Constants

x = np.linalg.solve(A, b)
print(f"Solution: x={x[0]:.4f}, y={x[1]:.4f}")

# Verify
print(f"Verify: A @ x = {A @ x}")  # Should equal b
```

## 6.4 — Practical: Least Squares Linear Regression

```python
# Generate synthetic data: y = 3x + 7 + noise
np.random.seed(42)
n = 100
x = np.random.uniform(0, 10, n)
y = 3 * x + 7 + np.random.normal(0, 2, n)

# Set up for least squares: y = X @ beta
# X has columns [x, 1] (for slope and intercept)
X = np.column_stack([x, np.ones(n)])

# Normal equation: beta = (X'X)^(-1) X'y
beta = np.linalg.lstsq(X, y, rcond=None)[0]
slope, intercept = beta

print(f"True: y = 3.00x + 7.00")
print(f"Fit:  y = {slope:.2f}x + {intercept:.2f}")

# R² score
y_pred = X @ beta
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - y.mean()) ** 2)
r_squared = 1 - ss_res / ss_tot
print(f"R² = {r_squared:.4f}")
```
