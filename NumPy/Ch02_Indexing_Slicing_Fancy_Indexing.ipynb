## 2.1 — Basic Indexing and Slicing

```python
import numpy as np

# 1D slicing (same as Python lists)
a = np.arange(10)  # [0, 1, 2, ..., 9]
print(f"a[3]:      {a[3]}")       # 3
print(f"a[2:5]:    {a[2:5]}")     # [2, 3, 4]
print(f"a[::2]:    {a[::2]}")     # [0, 2, 4, 6, 8]
print(f"a[::-1]:   {a[::-1]}")    # [9, 8, ..., 0]

# 2D indexing
m = np.arange(20).reshape(4, 5)
print(f"\nMatrix:\n{m}")
print(f"m[1, 3]:     {m[1, 3]}")       # Element at row 1, col 3
print(f"m[1]:        {m[1]}")           # Entire row 1
print(f"m[:, 2]:     {m[:, 2]}")        # Entire column 2
print(f"m[1:3, 2:4]:\n{m[1:3, 2:4]}")  # Submatrix
```

## 2.2 — Boolean Masking (Filtering)

```python
data = np.array([15, 22, 8, 35, 12, 42, 5, 28])

# Create a boolean mask
mask = data > 20
print(f"Mask: {mask}")           # [False, True, False, True, ...]
print(f"Filtered: {data[mask]}")  # [22, 35, 42, 28]

# Combine conditions (use & for AND, | for OR, ~ for NOT)
# ⚠️ Corner Case: MUST use & | ~ (not 'and' 'or' 'not')!
# ⚠️ Corner Case: MUST wrap each condition in parentheses!
result = data[(data > 10) & (data < 30)]
print(f"Between 10-30: {result}")

# Practical: replace values
cleaned = data.copy()
cleaned[cleaned < 10] = 0
print(f"Cleaned (< 10 → 0): {cleaned}")
```

## 2.3 — Fancy Indexing

```python
a = np.arange(0, 50, 5)  # [0, 5, 10, 15, 20, 25, 30, 35, 40, 45]

# Select specific indices
indices = [1, 4, 7]
print(f"Fancy index: {a[indices]}")  # [5, 20, 35]

# 2D fancy indexing
m = np.arange(12).reshape(3, 4)
print(f"\nMatrix:\n{m}")

# Select specific (row, col) pairs
rows = [0, 1, 2]
cols = [1, 2, 3]
print(f"Elements at (0,1),(1,2),(2,3): {m[rows, cols]}")  # [1, 6, 11]
```

## 2.4 — `np.where()` — Conditional Selection

```python
a = np.array([1, -2, 3, -4, 5, -6])

# np.where(condition, value_if_true, value_if_false)
result = np.where(a > 0, a, 0)  # Replace negatives with 0
print(f"Clipped negatives: {result}")  # [1, 0, 3, 0, 5, 0]

# np.where without values → returns indices
indices = np.where(a > 0)
print(f"Indices of positives: {indices}")

# Practical: categorize data
scores = np.array([85, 42, 92, 68, 75, 55])
grades = np.where(scores >= 70, "Pass", "Fail")
print(f"Grades: {grades}")
```

## 2.5 — ⚠️ Critical Corner Case: Views vs Copies

NumPy slicing returns a **view** (not a copy). Modifying the view modifies the original!

```python
original = np.array([1, 2, 3, 4, 5])

# Slicing creates a VIEW — shares memory!
view = original[1:4]
view[0] = 99
print(f"Original after modifying view: {original}")
# [1, 99, 3, 4, 5] — CHANGED!

# ✅ Use .copy() for an independent copy
original2 = np.array([1, 2, 3, 4, 5])
independent = original2[1:4].copy()
independent[0] = 99
print(f"Original after modifying copy: {original2}")
# [1, 2, 3, 4, 5] — Unchanged!

# How to check if it's a view:
a = np.arange(10)
s = a[2:5]
print(f"\nIs s a view? {s.base is not None}")   # True
print(f"s shares memory with a? {np.shares_memory(a, s)}")

# ⚠️ Fancy indexing and boolean indexing return COPIES, not views!
fancy = a[[1, 3, 5]]
fancy[0] = 99
print(f"After fancy index modify: {a}")  # Unchanged!
```
