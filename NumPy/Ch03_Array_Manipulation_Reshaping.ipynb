## 3.1 — Reshaping

```python
import numpy as np

a = np.arange(12)
print(f"Original: {a}")

# reshape — returns a view (if possible)
r = a.reshape(3, 4)
print(f"Reshaped (3,4):\n{r}")

# Use -1 to auto-calculate one dimension
print(f"reshape(2, -1):\n{a.reshape(2, -1)}")   # (2, 6)
print(f"reshape(-1, 3):\n{a.reshape(-1, 3)}")   # (4, 3)

# ⚠️ ravel() vs flatten()
# ravel → returns a VIEW (fast, but changes affect original!)
# flatten → returns a COPY (safe, but uses extra memory)
r1 = r.ravel()
r2 = r.flatten()

r1[0] = 99
print(f"\nAfter ravel modify — original changed: {r[0, 0]}")  # 99
r2[0] = 88
print(f"After flatten modify — original unchanged: {r[0, 0]}")  # 99 (unchanged from last)
```

## 3.2 — Transpose

```python
m = np.arange(6).reshape(2, 3)
print(f"Original (2,3):\n{m}")

# Transpose — swap rows and columns
print(f"\nTranspose (3,2):\n{m.T}")

# For higher dimensions: swapaxes
a3d = np.arange(24).reshape(2, 3, 4)
print(f"\n3D shape: {a3d.shape}")
print(f"swapaxes(0,2): {a3d.swapaxes(0, 2).shape}")  # (4, 3, 2)
print(f"transpose(2,0,1): {a3d.transpose(2, 0, 1).shape}")  # (4, 2, 3)
```

## 3.3 — Stacking & Concatenation

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

# Vertical stack (add rows)
print(f"vstack:\n{np.vstack([a, b])}")   # (4, 2)

# Horizontal stack (add columns)
print(f"\nhstack:\n{np.hstack([a, b])}")  # (2, 4)

# concatenate — general version (specify axis)
print(f"\nconcat axis=0:\n{np.concatenate([a, b], axis=0)}")  # Same as vstack
print(f"\nconcat axis=1:\n{np.concatenate([a, b], axis=1)}")  # Same as hstack

# stack — creates a NEW dimension
stacked = np.stack([a, b], axis=0)  # (2, 2, 2)
print(f"\nstack shape: {stacked.shape}")
print(f"stack:\n{stacked}")
```

## 3.4 — Splitting

```python
a = np.arange(16).reshape(4, 4)
print(f"Original:\n{a}\n")

# Split into equal parts
top, bottom = np.vsplit(a, 2)
print(f"Top half:\n{top}")
print(f"Bottom half:\n{bottom}")

left, right = np.hsplit(a, 2)
print(f"\nLeft half:\n{left}")
print(f"Right half:\n{right}")

# Split at specific indices
parts = np.split(np.arange(10), [3, 7])
print(f"\nSplit at [3,7]: {[p.tolist() for p in parts]}")
```

## 3.5 — Adding & Removing Dimensions

```python
a = np.array([1, 2, 3])  # shape: (3,)

# Add dimension with np.newaxis
row = a[np.newaxis, :]   # (1, 3) — row vector
col = a[:, np.newaxis]   # (3, 1) — column vector
print(f"Row vector: {row.shape}")
print(f"Col vector: {col.shape}")

# expand_dims (same thing, clearer intent)
expanded = np.expand_dims(a, axis=0)  # (1, 3)

# squeeze — remove dimensions of size 1
squeezed = np.squeeze(row)  # Back to (3,)
print(f"Squeezed: {squeezed.shape}")

# tile and repeat
print(f"\ntile (1→3 copies):\n{np.tile(a, (2, 3))}")
print(f"\nrepeat (each elem 3x): {np.repeat(a, 3)}")
```
