## 7.1 — Classic Random Functions

```python
import numpy as np

np.random.seed(42)  # Reproducibility

print("Uniform [0,1):", np.random.rand(5))
print("Normal μ=0,σ=1:", np.random.randn(5).round(3))
print("Integers [0,10):", np.random.randint(0, 10, 5))
print("Choice:", np.random.choice(['a', 'b', 'c', 'd'], 3))
print("Choice weighted:", np.random.choice(
    ['common', 'rare'], 5, p=[0.9, 0.1]))

# Shuffle (in-place!)
a = np.arange(10)
np.random.shuffle(a)
print(f"Shuffled: {a}")

# Permutation (returns new array)
perm = np.random.permutation(10)
print(f"Permutation: {perm}")
```

## 7.2 — Modern Generator API (Preferred!)

The new API uses `default_rng()` — more predictable, better statistical properties, and thread-safe.

```python
rng = np.random.default_rng(seed=42)

print("Uniform:", rng.random(5).round(3))
print("Normal:", rng.standard_normal(5).round(3))
print("Integers:", rng.integers(0, 10, 5))
print("Choice:", rng.choice(['a', 'b', 'c'], 3))

# Distributions
print("\nBinomial (n=10, p=0.5):", rng.binomial(10, 0.5, 5))
print("Poisson (λ=3):", rng.poisson(3, 5))
print("Exponential (β=2):", rng.exponential(2, 5).round(3))
```

## 7.3 — Statistical Functions

```python
data = np.random.default_rng(42).normal(50, 15, 1000)  # μ=50, σ=15

print(f"Mean:       {data.mean():.2f}")
print(f"Median:     {np.median(data):.2f}")
print(f"Std Dev:    {data.std():.2f}")
print(f"Variance:   {data.var():.2f}")
print(f"Min:        {data.min():.2f}")
print(f"Max:        {data.max():.2f}")

# Percentiles
print(f"\n25th percentile: {np.percentile(data, 25):.2f}")
print(f"50th percentile: {np.percentile(data, 50):.2f}")
print(f"75th percentile: {np.percentile(data, 75):.2f}")

# Correlation
x = np.random.default_rng(42).normal(0, 1, 100)
y = 2 * x + np.random.default_rng(43).normal(0, 0.5, 100)
corr = np.corrcoef(x, y)
print(f"\nCorrelation matrix:\n{corr.round(3)}")

# Histogram
counts, bin_edges = np.histogram(data, bins=10)
print(f"\nHistogram counts: {counts}")
print(f"Bin edges: {bin_edges.round(1)}")
```

## 7.4 — Practical: Monte Carlo Estimation of π

```python
# Method: Random points in a unit square [0,1] × [0,1]
# If x² + y² ≤ 1, the point is inside the quarter circle.
# Ratio ≈ π/4

rng = np.random.default_rng(42)

for n in [100, 1_000, 10_000, 100_000, 1_000_000]:
    x = rng.random(n)
    y = rng.random(n)
    inside = np.sum(x**2 + y**2 <= 1)
    pi_estimate = 4 * inside / n
    error = abs(pi_estimate - np.pi)
    print(f"n={n:>10,}: π ≈ {pi_estimate:.6f}  (error={error:.6f})")
```
