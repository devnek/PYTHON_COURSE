## 1.1 — What is NumPy?

**NumPy** (Numerical Python) is the foundational package for scientific computing in Python. It provides:
- **ndarray**: A fast, memory-efficient N-dimensional array
- **Vectorized operations**: Operate on entire arrays without Python loops
- **Broadcasting**: Automatic shape matching for arithmetic
- **Linear algebra, FFT, random**: Built-in mathematical tools

### Why is NumPy fast?
1. Arrays are stored in **contiguous memory** (unlike Python lists)
2. Operations are implemented in **C/Fortran** under the hood
3. **Vectorization** eliminates Python loop overhead

```python
import numpy as np

# Speed comparison: NumPy vs Python lists
import time

size = 1_000_000

# Python list
py_list = list(range(size))
start = time.time()
py_result = [x * 2 for x in py_list]
py_time = time.time() - start

# NumPy array
np_array = np.arange(size)
start = time.time()
np_result = np_array * 2
np_time = time.time() - start

print(f"Python list: {py_time:.4f}s")
print(f"NumPy array: {np_time:.4f}s")
print(f"NumPy is {py_time/np_time:.0f}x faster!")
```

## 1.2 — Creating Arrays

```python
import numpy as np

# From Python data
a1 = np.array([1, 2, 3, 4, 5])               # 1D
a2 = np.array([[1, 2, 3], [4, 5, 6]])         # 2D
a3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])  # 3D

# Special arrays
zeros = np.zeros((3, 4))                 # 3x4 of zeros
ones = np.ones((2, 3))                   # 2x3 of ones
full = np.full((2, 3), 7)               # 2x3 filled with 7
eye = np.eye(4)                          # 4x4 identity matrix
empty = np.empty((2, 2))                # Uninitialized (fast, but garbage values!)

# Sequences
arange = np.arange(0, 10, 2)            # [0, 2, 4, 6, 8] — like range()
linspace = np.linspace(0, 1, 5)         # [0, 0.25, 0.5, 0.75, 1.0] — N evenly spaced

# Random
rand_uniform = np.random.rand(3, 3)     # Uniform [0, 1)
rand_normal = np.random.randn(3, 3)     # Normal (mean=0, std=1)
rand_int = np.random.randint(0, 10, (3, 3))  # Integers [0, 10)

print("zeros:\n", zeros)
print("\narange:", arange)
print("linspace:", linspace)
```

## 1.3 — Array Attributes

```python
a = np.random.randn(3, 4, 5)

print(f"Shape:        {a.shape}")      # (3, 4, 5)
print(f"Dimensions:   {a.ndim}")       # 3
print(f"Data type:    {a.dtype}")      # float64
print(f"Total size:   {a.size}")       # 60 (3*4*5)
print(f"Item size:    {a.itemsize} bytes")  # 8 (float64 = 8 bytes)
print(f"Total memory: {a.nbytes} bytes")   # 480 (60 * 8)
```

## 1.4 — Data Types (`dtype`)

```python
# NumPy has many specific dtypes
int_arr = np.array([1, 2, 3])               # int64 (default)
float_arr = np.array([1.0, 2.0, 3.0])      # float64 (default)
bool_arr = np.array([True, False, True])    # bool
str_arr = np.array(["a", "b", "c"])         # <U1 (Unicode string)

# Specify dtype explicitly
a = np.array([1, 2, 3], dtype=np.float32)   # Force float32
b = np.array([1.9, 2.8, 3.7], dtype=np.int32)  # Truncates! → [1, 2, 3]
print(f"Float32: {a}, dtype={a.dtype}")
print(f"Truncated: {b}")

# Cast dtype
c = a.astype(np.int64)
print(f"Cast to int64: {c}")

# ⚠️ Corner Case: float16 precision
x = np.array([65536.0], dtype=np.float16)
print(f"float16 overflow: {x}")  # inf! max float16 ≈ 65504
```
