## 3.1 — String Creation

````python
# Single or double quotes — no difference
s1 = 'hello'
s2 = "hello"

# Triple quotes for multi-line strings
s3 = """This is
a multi-line
string."""
print(s3)

# Raw strings — backslashes are literal (great for regex, file paths)
path = r"C:\Users\name\folder"
print(path)  # C:\Users\name\folder  (no escaping!)
````

## 3.2 — Indexing and Slicing
```
 P  y  t  h  o  n
 0  1  2  3  4  5     ← positive index
-6 -5 -4 -3 -2 -1    ← negative index
```

````python
s = "Python"

# Indexing
print(s[0])    # P
print(s[-1])   # n  (last character)

# Slicing: s[start:stop:step]
print(s[0:3])    # Pyt  (stop is exclusive)
print(s[2:])     # thon (from index 2 to end)
print(s[:3])     # Pyt  (from start to index 3)
print(s[::2])    # Pto  (every 2nd character)

# 🔥 TRICK: Reverse a string
print(s[::-1])   # nohtyP
````

## 3.3 — Essential String Methods

````python
text = "  Hello, World!  "

# Whitespace
print(text.strip())    # "Hello, World!" — removes leading/trailing whitespace
print(text.lstrip())   # "Hello, World!  "
print(text.rstrip())   # "  Hello, World!"

# Case
print("hello".upper())        # HELLO
print("HELLO".lower())        # hello
print("hello world".title())  # Hello World
print("hello world".capitalize())  # Hello world
print("Hello".swapcase())     # hELLO

# Search
print("banana".count("an"))   # 2
print("banana".find("na"))    # 2 (index of first occurrence)
print("banana".find("xyz"))   # -1 (not found)
print("banana".startswith("ban"))  # True
print("banana".endswith("ana"))    # True

# Replace
print("Hello World".replace("World", "Python"))  # Hello Python

# Split and Join
csv = "apple,banana,cherry"
fruits = csv.split(",")
print(fruits)                         # ['apple', 'banana', 'cherry']
print(" → ".join(fruits))             # apple → banana → cherry
````

## 3.4 — 🔥 String Tricks

````python
# Trick 1: Check if string contains only digits/letters
print("12345".isdigit())     # True
print("abc123".isalnum())    # True
print("   ".isspace())       # True

# Trick 2: String multiplication
print("=" * 40)    # ========================================
print("ha" * 3)    # hahaha

# Trick 3: Check for palindrome
word = "racecar"
print(word == word[::-1])  # True — it's a palindrome!

# Trick 4: Center / pad strings
print("TITLE".center(30, "="))   # ============TITLE=============
print("42".zfill(5))              # 00042  — zero-padding!

# Trick 5: Translate characters (like a cipher!)
table = str.maketrans("aeiou", "12345")
print("hello world".translate(table))  # h2ll4 w4rld

# Trick 6: Partition — split into 3 parts
email = "user@example.com"
name, sep, domain = email.partition("@")
print(f"Name: {name}, Domain: {domain}")

# Trick 7: Multi-line string join (cleaner than concatenation)
query = " ".join([
    "SELECT *",
    "FROM users",
    "WHERE age > 25",
    "ORDER BY name",
])
print(query)
````
