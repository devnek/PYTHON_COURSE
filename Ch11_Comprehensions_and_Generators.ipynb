## 11.1 — List Comprehensions
`[expression for item in iterable if condition]`

```python
# Basic
squares = [x**2 for x in range(10)]
print(squares)

# With condition
evens = [x for x in range(20) if x % 2 == 0]
print(evens)

# With if/else (note: no 'if' at the end — it goes BEFORE 'for')
labels = ["even" if x % 2 == 0 else "odd" for x in range(6)]
print(labels)  # ['even', 'odd', 'even', 'odd', 'even', 'odd']

# 🔥 Trick: Flatten + filter in one comprehension
matrix = [[1, -2, 3], [-4, 5, -6], [7, -8, 9]]
positive_flat = [n for row in matrix for n in row if n > 0]
print(positive_flat)  # [1, 3, 5, 7, 9]
```

## 11.2 — Dict and Set Comprehensions

```python
# Dict comprehension
word = "abracadabra"
char_freq = {ch: word.count(ch) for ch in set(word)}
print("Frequencies:", char_freq)

# Set comprehension
sentence = "the quick brown fox jumps over the lazy dog"
unique_lengths = {len(word) for word in sentence.split()}
print("Unique word lengths:", sorted(unique_lengths))

# 🔥 Nested dict comprehension — multiplication table
table = {i: {j: i*j for j in range(1, 6)} for i in range(1, 6)}
for row, cols in table.items():
    print(f"{row}: {cols}")
```

## 11.3 — Generator Expressions
Like list comprehensions, but **lazy** — they compute values on demand.
Use `()` instead of `[]`.

```python
# List comprehension — creates entire list in memory
list_comp = [x**2 for x in range(1_000_000)]

# Generator expression — creates values one at a time (memory efficient!)
gen_expr = (x**2 for x in range(1_000_000))

print(type(list_comp))  # <class 'list'>
print(type(gen_expr))   # <class 'generator'>

# Generators work with sum(), max(), min(), any(), all()
total = sum(x**2 for x in range(1_000_000))  # No extra [] needed!
print(f"Sum of squares: {total}")

import sys
# Memory comparison
list_size = sys.getsizeof([x for x in range(10000)])
gen_size = sys.getsizeof(x for x in range(10000))
print(f"List: {list_size:,} bytes vs Generator: {gen_size} bytes")
```

## 11.4 — Generator Functions with `yield`

```python
def countdown(n):
    """A generator that counts down from n to 1."""
    while n > 0:
        yield n   # Pause here, return value, resume later
        n -= 1

# Use it
for num in countdown(5):
    print(num, end=" ")  # 5 4 3 2 1
print()

# Generators are exhaustible — can only iterate once!
gen = countdown(3)
print(list(gen))  # [3, 2, 1]
print(list(gen))  # [] — empty! Already exhausted.
```

## 11.5 — 🔥 Practical Generator: Reading Large Files Line by Line

```python
def read_large_file(filepath):
    """Read a file lazily, one line at a time."""
    with open(filepath, 'r') as f:
        for line in f:
            yield line.strip()

# Create a sample file to demonstrate
with open("sample.txt", "w") as f:
    for i in range(10):
        f.write(f"Line {i}: This is sample data\n")

# Read lazily — never loads the whole file into memory!
for line in read_large_file("sample.txt"):
    if "5" in line:
        print(f"Found: {line}")
        break  # Stop early — remaining lines are never read!
```

## 11.6 — 🔥 Generator Pipeline Pattern

```python
# Chain generators together like Unix pipes!
def numbers():
    yield from range(1, 21)  # yield from delegates to another iterable

def evens(source):
    for n in source:
        if n % 2 == 0:
            yield n

def squared(source):
    for n in source:
        yield n ** 2

# Pipeline: numbers → filter evens → square them
pipeline = squared(evens(numbers()))
result = list(pipeline)
print(result)  # [4, 16, 36, 64, 100, 144, 196, 256, 324, 400]
```
