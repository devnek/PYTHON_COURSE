## 9.1 — The `for` Loop

```python
# Iterate over any iterable
for fruit in ["apple", "banana", "cherry"]:
    print(fruit)

# range(start, stop, step)
for i in range(0, 10, 2):
    print(i, end=" ")  # 0 2 4 6 8
print()
```

## 9.2 — 🔥 Trick: NEVER use `range(len(...))`!
Use `enumerate()` instead.

```python
fruits = ["apple", "banana", "cherry"]

# ❌ Bad (C-style thinking)
for i in range(len(fruits)):
    print(f"{i}: {fruits[i]}")

print("---")

# ✅ Pythonic!
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")

# Start from a different number
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")
```

## 9.3 — 🔥 `zip()` — Loop Multiple Iterables Together

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
grades = ["B", "A", "C"]

# Zip three lists together
for name, score, grade in zip(names, scores, grades):
    print(f"{name}: {score} ({grade})")

# Create a dict from two lists
name_score = dict(zip(names, scores))
print(name_score)

# 🔥 Trick: Unzip with zip(*...)
pairs = [(1, 'a'), (2, 'b'), (3, 'c')]
numbers, letters = zip(*pairs)
print(f"Numbers: {numbers}, Letters: {letters}")

# zip_longest from itertools (when lists have different lengths)
from itertools import zip_longest
a = [1, 2, 3]
b = ['a', 'b']
print(list(zip_longest(a, b, fillvalue='?')))  # [(1,'a'), (2,'b'), (3,'?')]
```

## 9.4 — `while` Loops and the Walrus Operator

```python
# Classic while
count = 0
while count < 5:
    print(count, end=" ")
    count += 1
print()

# 🔥 Walrus operator := (Python 3.8+)
# Assign AND test in one expression
import random
random.seed(42)
while (roll := random.randint(1, 6)) != 6:
    print(f"Rolled {roll}, trying again...")
print(f"Rolled {roll}! Got a 6!")
```

## 9.5 — `break`, `continue`, and `for...else`

```python
# break — exit the loop
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")  # 0 1 2 3 4
print()

# continue — skip to next iteration
for i in range(10):
    if i % 3 == 0:
        continue
    print(i, end=" ")  # 1 2 4 5 7 8
print()

# 🔥 for...else — else runs only if loop completed WITHOUT break
# Great for search patterns!
target = 7
for num in [2, 4, 6, 7, 8]:
    if num == target:
        print(f"Found {target}!")
        break
else:
    print(f"{target} not found.")  # Only runs if break was NOT hit
```

## 9.6 — 🔥 Advanced Iteration Tricks

```python
# Trick 1: Iterate in reverse
for x in reversed(range(5)):
    print(x, end=" ")  # 4 3 2 1 0
print()

# Trick 2: Iterate over a dict sorted by value
scores = {"Alice": 85, "Bob": 92, "Charlie": 78}
for name, score in sorted(scores.items(), key=lambda x: x[1], reverse=True):
    print(f"{name}: {score}")

# Trick 3: any() and all() — check conditions across iterables
nums = [2, 4, 6, 8]
print(f"All even? {all(n % 2 == 0 for n in nums)}")  # True
print(f"Any > 5?  {any(n > 5 for n in nums)}")        # True

# Trick 4: Iterate with a sliding window
data = [1, 2, 3, 4, 5, 6]
window_size = 3
windows = [data[i:i+window_size] for i in range(len(data) - window_size + 1)]
print("Sliding windows:", windows)  # [[1,2,3], [2,3,4], [3,4,5], [4,5,6]]
```
