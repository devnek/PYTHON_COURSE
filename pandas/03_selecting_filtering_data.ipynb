# Chapter 3: Selecting & Filtering Data

This chapter covers every way to select and filter data in Pandas:

1. Column selection
2. Row selection with `.loc` and `.iloc`
3. Boolean filtering / masking
4. `.query()` method
5. `.isin()`, `.between()`, `.str.contains()`
6. Chained indexing pitfall

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace', 'Hank'],
    'department': ['Engineering', 'Marketing', 'Engineering', 'HR',
                   'Marketing', 'Engineering', 'HR', 'Marketing'],
    'salary': [95000, 65000, 110000, 72000, 68000, 105000, 78000, 62000],
    'years_exp': [5, 3, 8, 4, 2, 7, 6, 1],
    'rating': [4.5, 3.8, 4.9, 4.2, 3.5, 4.7, 4.0, 3.2],
})
df
```

## 3.1 Column Selection

```python
# Single column → returns a Series
print(type(df['name']))
df['name']
```

```python
# Multiple columns → returns a DataFrame
df[['name', 'salary']]
```

```python
# Dot notation (only for valid Python identifiers, no spaces)
df.salary.head(3)
```

## 3.2 Row Selection: `.loc` vs `.iloc`

| Accessor | Selection by | Endpoint |
|----------|-------------|----------|
| `.loc`   | **Labels**  | Inclusive |
| `.iloc`  | **Position** (integer) | Exclusive |

```python
# .loc — label-based (row labels are 0-7 here, which happen to match positions)
print('Single row:')
print(df.loc[0])
print()

# Rows 1 to 3 (INCLUSIVE), specific columns
df.loc[1:3, ['name', 'department', 'salary']]
```

```python
# .iloc — integer position-based
print('Row at position 0:')
print(df.iloc[0])
print()

# Rows 1 to 3 (EXCLUSIVE of 3), columns at positions 0 and 2
df.iloc[1:3, [0, 2]]
```

```python
# Fancy indexing with .iloc
df.iloc[[0, 3, 7], :]  # specific rows, all columns
```

## 3.3 Boolean Filtering (Masking)

```python
# Single condition
high_salary = df[df['salary'] > 80000]
high_salary
```

```python
# Multiple conditions — use & (and), | (or), ~ (not)
# IMPORTANT: each condition must be in parentheses!
eng_high = df[(df['department'] == 'Engineering') & (df['salary'] > 100000)]
eng_high
```

```python
# Negation with ~
not_marketing = df[~(df['department'] == 'Marketing')]
not_marketing
```

## 3.4 The `.query()` Method

A cleaner way to write filters using a string expression.

```python
# Same as: df[(df['salary'] > 80000) & (df['years_exp'] >= 5)]
df.query('salary > 80000 and years_exp >= 5')
```

```python
# Using variables with @
min_salary = 70000
dept = 'Engineering'

df.query('salary >= @min_salary and department == @dept')
```

## 3.5 `.isin()`, `.between()`, `.str.contains()`

```python
# .isin() — check membership in a list
target_depts = ['Engineering', 'HR']
df[df['department'].isin(target_depts)]
```

```python
# .between() — inclusive range check
df[df['salary'].between(70000, 100000)]
```

```python
# .str.contains() — substring / regex matching
df[df['name'].str.contains('a', case=False)]  # names containing 'a'
```

```python
# .str.startswith() / .str.endswith()
df[df['name'].str.startswith('C')]
```

## 3.6 ⚠️ Chained Indexing Pitfall

**Never** modify data using chained indexing — it may silently fail!

```python
# ❌ BAD — chained indexing (may trigger SettingWithCopyWarning)
# df[df['department'] == 'HR']['salary'] = 99999   # Don't do this!

# ✅ GOOD — use .loc for assignment
df_copy = df.copy()
df_copy.loc[df_copy['department'] == 'HR', 'salary'] = 99999
df_copy[df_copy['department'] == 'HR']
```

### Why Does This Happen?

Chained indexing like `df[condition][column]` creates an **intermediate copy**,  
so the assignment modifies the copy — not the original DataFrame.

**Rule of thumb**: Always use `.loc` for setting values.

## 📝 Key Takeaways

| Method | Use Case |
|--------|----------|
| `df['col']` | Select one column as Series |
| `df[['c1','c2']]` | Select multiple columns as DataFrame |
| `.loc[row, col]` | Select by **labels** (inclusive) |
| `.iloc[row, col]` | Select by **position** (exclusive end) |
| Boolean mask | Filter rows by conditions |
| `.query()` | Cleaner filter syntax |
| `.isin()` | Check membership in a list |
| `.between()` | Range filter |
| `.str.contains()` | Text pattern matching |
| **Always use `.loc`** | For modifying filtered data |
