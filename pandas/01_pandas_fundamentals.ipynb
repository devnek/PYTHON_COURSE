## 1.2 Series

A **Series** is a one-dimensional labeled array capable of holding any data type.

### Creating a Series

````python
# From a list
s1 = pd.Series([10, 20, 30, 40])
print('From list:\n', s1)
print()

# With custom index
s2 = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print('Custom index:\n', s2)
print()

# From a dictionary
s3 = pd.Series({'x': 3.14, 'y': 2.72, 'z': 1.41})
print('From dict:\n', s3)
print()

# From a NumPy array
s4 = pd.Series(np.arange(5), name='my_series')
print('From NumPy:\n', s4)
````

### Indexing a Series

````python
# Label-based indexing
print('s2["b"]  =', s2['b'])

# Position-based indexing
print('s2.iloc[1] =', s2.iloc[1])

# Slicing (label-based is INCLUSIVE on both ends)
print('\nSlice a:b\n', s2.loc['a':'b'])

# Boolean indexing
print('\nValues > 150:\n', s2[s2 > 150])
````

### Vectorized Operations

````python
temps_c = pd.Series([0, 20, 37, 100], index=['freezing', 'room', 'body', 'boiling'])

# Vectorized arithmetic — no loops needed!
temps_f = temps_c * 9/5 + 32
print('Celsius to Fahrenheit:\n', temps_f)

# NumPy ufuncs work directly on Series
print('\nSquare root:\n', np.sqrt(temps_c))
````

### Series Attributes & Methods

````python
s = pd.Series([10, 20, np.nan, 40, 50], name='demo')

print('name  :', s.name)
print('dtype :', s.dtype)
print('size  :', s.size)
print('index :', s.index.tolist())
print('values:', s.values)
print('hasnans:', s.hasnans)
print()
print('--- Descriptive stats ---')
print(s.describe())
````

## 1.3 DataFrame

A **DataFrame** is a two-dimensional, size-mutable, tabular data structure with labeled axes (rows and columns).

### Creating a DataFrame from a Dictionary

````python
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'age': [25, 30, 35, 28],
    'city': ['New York', 'London', 'Paris', 'Tokyo'],
    'salary': [70000, 80000, 90000, 75000]
}
df = pd.DataFrame(data)
df
````

### Creating from a List of Dicts (row-oriented)

````python
rows = [
    {'product': 'Widget', 'price': 9.99, 'qty': 100},
    {'product': 'Gadget', 'price': 24.99, 'qty': 50},
    {'product': 'Doohickey', 'price': 4.99, 'qty': 200},
]
pd.DataFrame(rows)
````

### Creating from a 2-D NumPy Array

````python
arr = np.random.randint(60, 100, size=(5, 4))
subjects = ['Math', 'Science', 'English', 'History']
students = ['S1', 'S2', 'S3', 'S4', 'S5']

grades = pd.DataFrame(arr, columns=subjects, index=students)
grades
````

## 1.4 Viewing Data

Pandas provides many convenience methods to quickly inspect a DataFrame.

````python
# Build a slightly larger DataFrame for demonstration
np.random.seed(42)
n = 50
df_large = pd.DataFrame({
    'id': range(1, n + 1),
    'value': np.random.randn(n),
    'category': np.random.choice(['A', 'B', 'C'], n),
    'date': pd.date_range('2024-01-01', periods=n, freq='D'),
})
````

````python
# First 5 rows
df_large.head()
````

````python
# Last 3 rows
df_large.tail(3)
````

````python
# Shape (rows, cols)
print('Shape:', df_large.shape)
````

````python
# Column names
print('Columns:', df_large.columns.tolist())
````

````python
# Data types
print(df_large.dtypes)
````

````python
# Index
print('Index:', df_large.index)
````

````python
# Concise summary
df_large.info()
````

````python
# Statistical summary
df_large.describe()
````

````python
# Include non-numeric columns
df_large.describe(include='all')
````

## 1.5 Setting & Resetting the Index

````python
# Set a column as the index
df_indexed = df.set_index('name')
df_indexed
````

````python
# Reset back to default integer index
df_indexed.reset_index()
````

````python
# drop=True discards the old index entirely
df_indexed.reset_index(drop=True)
````

````python
# Multi-level index
df_multi = df.set_index(['city', 'name'])
df_multi
````

## 📝 Key Takeaways

| Concept | Key Point |
|---------|----------|
| **Series** | 1-D labeled array; like a single column |
| **DataFrame** | 2-D labeled table; collection of Series |
| **Viewing** | `head()`, `tail()`, `info()`, `describe()` for quick inspection |
| **Index** | Use `set_index()` / `reset_index()` to manage row labels |
| **Vectorization** | Operate on entire Series/DataFrames — avoid Python loops |
