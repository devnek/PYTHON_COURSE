## 6.1 — Tuple Basics
Tuples are **immutable** — once created, you can't change them.

```python
# Creating tuples
t1 = (1, 2, 3)
t2 = 1, 2, 3           # Parentheses are optional!
t3 = (42,)              # Single element — NEED the comma!
t4 = tuple([1, 2, 3])   # From a list
empty = ()

print(t1, type(t1))
print(t3, type(t3))
print(type((42)))  # <class 'int'> — without comma, just parentheses!
```

## 6.2 — 🔥 Tuple Unpacking Tricks

```python
# Basic unpacking
point = (3, 7)
x, y = point
print(f"x={x}, y={y}")

# Star unpacking — grab the rest
first, *middle, last = [1, 2, 3, 4, 5]
print(f"first={first}, middle={middle}, last={last}")

# Ignore values with _
_, important, _ = ("skip", "keep", "skip")
print(important)

# Nested unpacking
data = ("Alice", (85, 92, 78))
name, (test1, test2, test3) = data
print(f"{name}: {test1}, {test2}, {test3}")

# Swap multiple variables at once
a, b, c = 1, 2, 3
a, b, c = c, a, b
print(f"a={a}, b={b}, c={c}")  # a=3, b=1, c=2
```

## 6.3 — Tuples as Dict Keys and Set Elements
Because tuples are immutable, they can be used where lists cannot.

```python
# Tuples as dictionary keys (lists can't do this!)
locations = {
    (40.7128, -74.0060): "New York",
    (51.5074, -0.1278): "London",
    (35.6762, 139.6503): "Tokyo",
}
print(locations[(40.7128, -74.0060)])

# Tuples in sets
visited = {(1, 2), (3, 4), (1, 2)}
print(visited)  # {(1, 2), (3, 4)} — duplicate removed
```

## 6.4 — `collections.namedtuple` — Tuples with Names!
Give your tuple fields meaningful names instead of index numbers.

```python
from collections import namedtuple

# Define a named tuple type
Point = namedtuple("Point", ["x", "y"])
Color = namedtuple("Color", "red green blue")  # String shorthand

# Create instances
p = Point(3, 7)
c = Color(255, 128, 0)

# Access by name OR index
print(f"Point: x={p.x}, y={p.y}")
print(f"Color: r={c[0]}, g={c[1]}, b={c[2]}")

# Convert to dict
print(p._asdict())  # {'x': 3, 'y': 7}

# Create from dict
data = {"x": 10, "y": 20}
p2 = Point(**data)
print(p2)

# Replace values (returns new tuple — still immutable!)
p3 = p._replace(x=100)
print(f"Original: {p}, New: {p3}")
```
