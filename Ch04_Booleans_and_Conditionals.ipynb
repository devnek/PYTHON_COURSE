## 4.1 — Truthiness in Python
Every object in Python has a truth value. Here's what's **Falsy**:

```python
# ALL of these are Falsy (evaluate to False)
falsy_values = [False, 0, 0.0, 0j, "", [], (), {}, set(), None, range(0)]

for val in falsy_values:
    print(f"{str(val):>10} → bool = {bool(val)}")

# EVERYTHING ELSE is Truthy
print(f"\n{'[0]':>10} → bool = {bool([0])}")   # True! The list is non-empty
print(f"{'" "':>10} → bool = {bool(' ')}")     # True! Space is a character
```

## 4.2 — `if / elif / else`

```python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

print(f"Score {score} → Grade {grade}")
```

## 4.3 — 🔥 Trick: Ternary (Conditional) Expression
One-liner `if/else` — great for simple assignments.

```python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)  # adult

# Nested ternary (use sparingly!)
x = 0
sign = "positive" if x > 0 else ("negative" if x < 0 else "zero")
print(sign)
```

## 4.4 — 🔥 Trick: Short-Circuit Evaluation
`and` returns the first Falsy value (or the last value if all Truthy).
`or` returns the first Truthy value (or the last value if all Falsy).

```python
# 'or' for defaults
username = ""
display_name = username or "Anonymous"
print(display_name)  # Anonymous

# 'and' for guarding
data = {"name": "Alice"}
# Safe access: returns None if 'address' key doesn't exist
result = data.get("address") and data["address"].upper()
print(result)  # None (short-circuits because .get returns None which is Falsy)

# Chaining
print(1 and 2 and 3)   # 3 (all truthy → last value)
print(1 and 0 and 3)   # 0 (first falsy value)
print(0 or "" or "hi") # "hi" (first truthy value)
```

## 4.5 — `in` and `not` operators

```python
# Membership testing with 'in'
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)      # True
print("grape" not in fruits)   # True

# Works with strings too
print("Py" in "Python")  # True

# Works with dicts (checks keys)
user = {"name": "Alice", "age": 30}
print("name" in user)     # True
print("email" in user)    # False
```

## 4.6 — Match-Case (Python 3.10+)
Structural pattern matching — Python's version of switch/case, but way more powerful.

```python
# Basic match-case
command = "quit"

match command:
    case "start":
        print("Starting...")
    case "stop":
        print("Stopping...")
    case "quit" | "exit":          # Multiple patterns with |
        print("Goodbye!")
    case _:                        # Default / wildcard
        print("Unknown command")
```
