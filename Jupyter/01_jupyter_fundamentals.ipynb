## 1.1 What is Jupyter?

### A Brief History

| Year | Milestone |
|------|-----------|
| **2001** | Fernando Pérez creates **IPython** — an enhanced interactive Python shell |
| **2011** | The **IPython Notebook** (browser-based) is introduced |
| **2014** | Project **Jupyter** is spun off from IPython |
| **2018** | **JupyterLab** — the next-generation UI — reaches 1.0 |
| **2020+** | Jupyter becomes the de-facto standard for data science workflows |

### Why "Jupyter"?

The name is a reference to the three core programming languages it originally supported:

- **Ju**lia
- **Pyt**hon
- **R**

It is also a tribute to Galileo's notebooks documenting the moons of **Jupiter**.

### What Makes Jupyter Special?

- **Interactive computing**: write & run code in small chunks
- **Narrative + code**: mix explanations, equations, and code in one document
- **Language agnostic**: 100+ kernels (Python, R, Julia, Scala, …)
- **Reproducibility**: share notebooks with code *and* results
- **Ecosystem**: integrates with pandas, matplotlib, scikit-learn, TensorFlow, …

## 1.2 Notebook Architecture

A Jupyter Notebook has **three main components**:

```
┌────────────────────────────────────────────┐
│              Web Browser (UI)              │
│  ┌──────────────────────────────────────┐  │
│  │  Cell 1  [Markdown]                  │  │
│  │  Cell 2  [Code]        ──────────┐   │  │
│  │  Cell 3  [Code]        ──────────┤   │  │
│  │  Cell 4  [Markdown]              │   │  │
│  └──────────────────────────────────┼───┘  │
│                                     │      │
│              HTTP / WebSocket        │      │
│                                     ▼      │
│  ┌──────────────────────────────────────┐  │
│  │        Jupyter Server                │  │
│  │   (manages files, sessions, …)       │  │
│  └──────────────┬───────────────────────┘  │
│                 │  ZeroMQ                   │
│                 ▼                           │
│  ┌──────────────────────────────────────┐  │
│  │          Kernel (Python)             │  │
│  │   (executes code, returns results)   │  │
│  └──────────────────────────────────────┘  │
└────────────────────────────────────────────┘
```

### Key Concepts

| Component | Role |
|-----------|------|
| **Kernel** | A separate process that executes code. Each notebook connects to one kernel. |
| **Cells** | The fundamental unit — each cell holds either code, markdown, or raw text. |
| **JSON format** | `.ipynb` files are plain JSON — you can inspect them in any text editor. |

### The `.ipynb` JSON Structure

```json
{
  "nbformat": 4,
  "metadata": { "kernelspec": { ... } },
  "cells": [
    {
      "cell_type": "code",
      "source": ["print('hello')"],
      "outputs": [...],
      "execution_count": 1
    }
  ]
}
```

Let's inspect this very notebook's JSON structure programmatically:

````python
import json

# Read our own notebook file to see the JSON structure
notebook_file = '01_jupyter_fundamentals.ipynb'

try:
    with open(notebook_file) as f:
        nb = json.load(f)
    
    print(f'nbformat: {nb["nbformat"]}.{nb["nbformat_minor"]}')
    print(f'Number of cells: {len(nb["cells"])}')
    print(f'Kernel: {nb["metadata"]["kernelspec"]["display_name"]}')
    print()
    
    # Show cell types breakdown
    from collections import Counter
    types = Counter(c['cell_type'] for c in nb['cells'])
    print('Cell type breakdown:')
    for ctype, count in types.most_common():
        print(f'  {ctype:10s} → {count} cells')
except FileNotFoundError:
    print(f'File {notebook_file} not found — run this from the Jupyter directory.')
````

## 1.3 Keyboard Shortcuts Cheat Sheet

Jupyter has **two modes**:

| Mode | How to Enter | Cell Border |
|------|-------------|-------------|
| **Command mode** | Press `Esc` | Blue |
| **Edit mode** | Press `Enter` or click inside cell | Green |

### Command Mode Shortcuts (press `Esc` first)

| Shortcut | Action |
|----------|--------|
| `A` | Insert cell **a**bove |
| `B` | Insert cell **b**elow |
| `D, D` | **D**elete selected cell |
| `Z` | Undo cell deletion |
| `M` | Change cell to **M**arkdown |
| `Y` | Change cell to Code |
| `R` | Change cell to **R**aw |
| `C` | **C**opy cell |
| `V` | Paste cell below |
| `X` | Cut cell |
| `Shift+M` | **M**erge selected cells |
| `L` | Toggle **l**ine numbers |
| `O` | Toggle **o**utput |
| `H` | Show keyboard **h**elp |
| `I, I` | **I**nterrupt kernel |
| `0, 0` | Restart kernel |

### Run Shortcuts (both modes)

| Shortcut | Action |
|----------|--------|
| `Shift+Enter` | Run cell, select below |
| `Ctrl+Enter` | Run cell, stay in cell |
| `Alt+Enter` | Run cell, insert below |

### Edit Mode Shortcuts

| Shortcut | Action |
|----------|--------|
| `Tab` | Code completion / indent |
| `Shift+Tab` | Tooltip / docstring |
| `Ctrl+Shift+-` | Split cell at cursor |
| `Ctrl+/` | Toggle comment |

## 1.4 Cell Types

Jupyter supports **three** cell types:

### 1. Code Cells
- Execute code via the kernel
- Display output directly below
- Show execution count `[n]` in the margin

````python
# This is a CODE cell
result = 2 ** 10
print(f'2^10 = {result}')
````

### 2. Markdown Cells
- Render formatted text when run
- Support headers, lists, bold, italic, links, images, tables
- Support LaTeX math equations
- Double-click a rendered markdown cell to edit it

### 3. Raw Cells
- Not executed and not rendered
- Useful for content that should pass through `nbconvert` unmodified
- Example: raw LaTeX for PDF export

````python
# Programmatically check cell types in this notebook
import json

try:
    with open('01_jupyter_fundamentals.ipynb') as f:
        nb = json.load(f)
    
    for i, cell in enumerate(nb['cells'][:5]):
        cell_type = cell['cell_type']
        preview = cell['source'][0][:60] if cell['source'] else '(empty)'
        print(f'Cell {i}: [{cell_type:8s}] {preview}...')
except FileNotFoundError:
    print('Run from the Jupyter directory to inspect the notebook.')
````

## 1.5 Magic Commands

IPython **magic commands** start with `%` (line magic) or `%%` (cell magic).

| Magic | Type | Purpose |
|-------|------|---------|
| `%timeit` | Line | Time a single statement (many runs) |
| `%%time` | Cell | Time the entire cell (single run) |
| `%who` | Line | List variables of a given type |
| `%whos` | Line | Detailed listing of variables |
| `%reset` | Line | Delete all variables from namespace |
| `%run` | Line | Run a Python script |
| `%load` | Line | Load a file's contents into a cell |
| `%lsmagic` | Line | List all available magics |

````python
# %timeit — benchmark a single line
# (Runs the statement many times and reports average)
%timeit sum(range(1000))
````

````python
%%time
# %%time — time the entire cell (single run)
total = 0
for i in range(1_000_000):
    total += i
print(f'Sum: {total:,}')
````

````python
# Create some variables for %who and %whos demos
name = 'Jupyter'
version = 4
pi_approx = 3.14159
languages = ['Python', 'Julia', 'R']
````

````python
# %who — list variables of a specific type
%who str
````

````python
# %whos — detailed variable listing
%whos
````

````python
# %lsmagic — see ALL available magics
%lsmagic
````

## 1.6 Shell Commands with `!`

Prefix any command with `!` to run it in the system shell:

```python
!ls -la          # list files
!pwd              # current directory
!pip list         # installed packages
!echo "Hello"     # any shell command
```

You can also **capture shell output** into Python variables:

````python
# Run shell commands from Jupyter
!echo 'Hello from the shell!'
!python --version
````

````python
# Capture shell output into a Python variable
files = !ls -1
print(f'Files in current directory: {len(files)}')
for f in files[:5]:
    print(f'  📄 {f}')
````
