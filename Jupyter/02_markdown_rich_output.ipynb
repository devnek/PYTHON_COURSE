## 2.1 Markdown Syntax

Markdown cells support standard Markdown plus GitHub-Flavored Markdown.

### Headers

```markdown
# H1 — Title
## H2 — Section
### H3 — Subsection
#### H4 — Sub-subsection
```

### Text Formatting

| Syntax | Result |
|--------|--------|
| `**bold**` | **bold** |
| `*italic*` | *italic* |
| `***bold italic***` | ***bold italic*** |
| `~~strikethrough~~` | ~~strikethrough~~ |
| `` `inline code` `` | `inline code` |

### Lists

**Unordered:**
- Item 1
- Item 2
  - Sub-item 2a
  - Sub-item 2b

**Ordered:**
1. First step
2. Second step
3. Third step

### Links and Images

```markdown
[Link text](https://jupyter.org)
![Alt text](path/to/image.png)
```

### Tables

```markdown
| Column A | Column B | Column C |
|----------|----------|----------|
| data 1   | data 2   | data 3   |
```

### Block Quotes

> "The purpose of computing is insight, not numbers."
> — Richard Hamming

### Horizontal Rules

Use `---`, `***`, or `___` for a horizontal rule.

## 2.2 LaTeX Math Equations

Jupyter renders LaTeX math using **MathJax**.

### Inline Math

Use single `$` delimiters: The area of a circle is $A = \pi r^2$.

### Block Math

Use double `$$` delimiters:

$$
E = mc^2
$$

### More Examples

**Quadratic formula:**

$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

**Summation:**

$$
\sum_{i=1}^{n} i = \frac{n(n+1)}{2}
$$

**Matrix:**

$$
\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}
\begin{bmatrix}
x \\
y
\end{bmatrix}
=
\begin{bmatrix}
ax + by \\
cx + dy
\end{bmatrix}
$$

**Integral:**

$$
\int_0^\infty e^{-x^2} dx = \frac{\sqrt{\pi}}{2}
$$

### Common LaTeX Symbols

| Symbol | LaTeX | Symbol | LaTeX |
|--------|-------|--------|-------|
| $\alpha$ | `\alpha` | $\beta$ | `\beta` |
| $\gamma$ | `\gamma` | $\delta$ | `\delta` |
| $\sum$ | `\sum` | $\prod$ | `\prod` |
| $\infty$ | `\infty` | $\partial$ | `\partial` |
| $\nabla$ | `\nabla` | $\approx$ | `\approx` |

## 2.4 The `IPython.display` Module

The `IPython.display` module provides functions for rich output in **code cells**:

| Function | Purpose |
|----------|---------|
| `display()` | Display any object with rich representation |
| `HTML()` | Render HTML |
| `Markdown()` | Render Markdown |
| `Image()` | Display images |
| `SVG()` | Display SVG graphics |
| `JSON()` | Interactive JSON viewer |
| `Latex()` | Render LaTeX |

````python
from IPython.display import display, HTML, Markdown, Latex
````

````python
# Display HTML from a code cell
html_content = """
<table style="border-collapse: collapse; width: 100%;">
  <tr style="background: #4CAF50; color: white;">
    <th style="padding: 12px;">Feature</th>
    <th style="padding: 12px;">Support</th>
  </tr>
  <tr style="background: #f2f2f2;">
    <td style="padding: 12px;">HTML Tables</td>
    <td style="padding: 12px;">✅ Full</td>
  </tr>
  <tr>
    <td style="padding: 12px;">CSS Styling</td>
    <td style="padding: 12px;">✅ Full</td>
  </tr>
  <tr style="background: #f2f2f2;">
    <td style="padding: 12px;">JavaScript</td>
    <td style="padding: 12px;">⚠️ Limited</td>
  </tr>
</table>
"""

display(HTML(html_content))
````

````python
# Display Markdown from a code cell — great for dynamic content!
import datetime

now = datetime.datetime.now()
md_content = f"""
## 📅 Dynamic Report

Generated on: **{now.strftime('%Y-%m-%d %H:%M:%S')}**

| Metric | Value |
|--------|-------|
| Day of week | {now.strftime('%A')} |
| Day of year | {now.timetuple().tm_yday} |
| Week number | {now.isocalendar()[1]} |
"""

display(Markdown(md_content))
````

````python
# Display LaTeX from a code cell
display(Latex(r'$\nabla \cdot \mathbf{E} = \frac{\rho}{\epsilon_0}$ (Gauss\'s Law)'))
display(Latex(r'$$\oint \mathbf{B} \cdot d\mathbf{l} = \mu_0 I + \mu_0 \epsilon_0 \frac{d\Phi_E}{dt}$$'))
````

## 2.5 Rich Output with `_repr_html_` and `_repr_latex_`

Custom classes can define special methods for Jupyter to render them richly:

| Method | Output format |
|--------|---------------|
| `_repr_html_()` | HTML table, styled output |
| `_repr_latex_()` | LaTeX equation |
| `_repr_markdown_()` | Markdown text |
| `_repr_json_()` | Interactive JSON |
| `_repr_png_()` | PNG image bytes |
| `_repr_svg_()` | SVG markup |

````python
class DataTable:
    """A simple table class with rich HTML representation."""
    
    def __init__(self, headers, rows, title=''):
        self.headers = headers
        self.rows = rows
        self.title = title
    
    def _repr_html_(self):
        """Rich HTML rendering for Jupyter."""
        html = '<div style="margin: 10px 0;">'
        if self.title:
            html += f'<h4 style="color: #2c3e50;">{self.title}</h4>'
        html += '<table style="border-collapse: collapse; width: auto;">'
        
        # Header row
        html += '<tr>'
        for h in self.headers:
            html += f'<th style="background: #3498db; color: white; padding: 8px 16px;">{h}</th>'
        html += '</tr>'
        
        # Data rows
        for i, row in enumerate(self.rows):
            bg = '#ecf0f1' if i % 2 == 0 else 'white'
            html += f'<tr style="background: {bg};">'
            for val in row:
                html += f'<td style="padding: 8px 16px; border: 1px solid #ddd;">{val}</td>'
            html += '</tr>'
        
        html += '</table></div>'
        return html
    
    def __repr__(self):
        """Fallback plain-text representation."""
        lines = [' | '.join(self.headers)]
        lines.append('-' * len(lines[0]))
        for row in self.rows:
            lines.append(' | '.join(str(v) for v in row))
        return '\n'.join(lines)


# Jupyter will automatically call _repr_html_()!
table = DataTable(
    headers=['Planet', 'Diameter (km)', 'Distance from Sun (AU)'],
    rows=[
        ['Mercury', '4,879', '0.39'],
        ['Venus', '12,104', '0.72'],
        ['Earth', '12,756', '1.00'],
        ['Mars', '6,792', '1.52'],
        ['Jupiter', '142,984', '5.20'],
    ],
    title='🪐 Inner Solar System'
)

table  # Just placing it at the end of a cell triggers rich display
````

````python
class Fraction:
    """A fraction class with LaTeX rendering."""
    
    def __init__(self, numerator, denominator):
        self.num = numerator
        self.den = denominator
    
    def _repr_latex_(self):
        return f'$$\\frac{{{self.num}}}{{{self.den}}}$$'
    
    def __repr__(self):
        return f'{self.num}/{self.den}'


# Jupyter renders this as a beautiful fraction
f = Fraction('x^2 + 1', '2x - 3')
f
````
