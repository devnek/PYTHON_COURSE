## 1. Introduction to Python

Python is a high-level, interpreted programming language known for its simplicity and readability. It supports multiple programming paradigms:
- **Procedural Programming**: Structuring code into step-by-step procedures and functions.
- **Object-Oriented Programming (OOP)**: Organizing state and behavior into classes and objects.
- **Functional Programming**: Treating computation as mathematical evaluations using functions, lambdas, and immutability.

### Key Features of Python
- **Easy to Read and Write**: Clean, indentation-based syntax designed to resemble readable pseudo-code.
- **Interpreted Language**: Code is executed line by line, making debugging interactive and quick.
- **Dynamic Typing**: Variables do not require explicit type declarations.
- **Extensive Standard Library**: "Batteries included" philosophy providing built-in tools for file management, math, data structures, and networking.
- **Vibrant Community**: A vast ecosystem of open-source frameworks and libraries.

## 3. What is a Virtual Environment?

A **virtual environment** is a self-contained directory tree that contains a specific Python installation alongside its own isolated packages and scripts.

### Why Virtual Environments are Essential
1. **Dependency Management**: Prevents packages from one project conflicting with another project.
2. **Version Control**: Allows Project A to use library version `1.0` while Project B uses version `2.0` on the same machine.
3. **System Isolation**: Keeps your global operating system Python installation clean and protected from accidental corruption.
4. **Reproducibility**: Easily recreate exact project dependencies using `pip freeze > requirements.txt`.

## 4. Creating and Using a Virtual Environment (`venv`)

Python 3 includes the native `venv` module by default — no third-party package needed.

### Step 1: Create a Virtual Environment
In your terminal, navigate to your project directory and run:
```bash
# Syntax: python3 -m venv <environment_name>
python3 -m venv .venv
```

### Step 2: Activate the Environment
- **On macOS / Linux**:
  ```bash
  source .venv/bin/activate
  ```
- **On Windows (PowerShell)**:
  ```powershell
  .\.venv\Scripts\Activate.ps1
  ```
- **On Windows (Command Prompt)**:
  ```cmd
  .\.venv\Scripts\activate.bat
  ```

*(When activated, your terminal prompt will typically show `(.venv)` as a prefix)*

### Step 3: Deactivate when Finished
```bash
deactivate
```

## 5. 🔥 Trick: Programmatically Checking Virtual Environment Status

You can inspect if a script or notebook is currently running inside an isolated virtual environment or using the system-wide Python installation.

````python
import sys

def check_virtual_env():
    # sys.base_prefix exists in Python 3.3+
    # In a virtual environment, sys.prefix points to the virtualenv folder,
    # while sys.base_prefix points to the underlying base Python installation.
    is_venv = getattr(sys, "base_prefix", sys.prefix) != sys.prefix or hasattr(sys, "real_prefix")
    
    print(f"In Virtual Environment: {is_venv}")
    print(f"Current sys.prefix:     {sys.prefix}")
    print(f"Base sys.base_prefix:   {getattr(sys, 'base_prefix', sys.prefix)}")

check_virtual_env()
````
