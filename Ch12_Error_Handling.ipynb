## 12.1 — Basic Try/Except

```python
# Without handling — program crashes
# print(10 / 0)  # ZeroDivisionError!

# With handling — program continues
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
    result = None

print(f"Result: {result}")
print("Program continues...")
```

## 12.2 — Try / Except / Else / Finally

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("  ❌ Division by zero!")
        return None
    except TypeError as e:
        print(f"  ❌ Type error: {e}")
        return None
    else:
        # Runs ONLY if no exception occurred
        print(f"  ✅ Success: {a}/{b} = {result}")
        return result
    finally:
        # ALWAYS runs — even if return was called above!
        print("  🔄 Cleanup done.")

print(safe_divide(10, 3))
print()
print(safe_divide(10, 0))
print()
print(safe_divide("10", 3))
```

## 12.3 — Common Built-in Exceptions

```python
exceptions_demo = {
    "ValueError": lambda: int("hello"),
    "KeyError": lambda: {"a": 1}["b"],
    "IndexError": lambda: [1, 2, 3][10],
    "AttributeError": lambda: "hello".nonexistent(),
    "FileNotFoundError": lambda: open("nonexistent.txt"),
    "TypeError": lambda: "hello" + 5,
}

for name, func in exceptions_demo.items():
    try:
        func()
    except Exception as e:
        print(f"{name:>20}: {e}")
```

## 12.4 — Raising and Custom Exceptions

```python
# Raising built-in exceptions
def validate_age(age):
    if not isinstance(age, int):
        raise TypeError("Age must be an integer")
    if age < 0 or age > 150:
        raise ValueError(f"Age must be 0-150, got {age}")
    return True

try:
    validate_age(200)
except ValueError as e:
    print(f"Validation failed: {e}")

# Custom Exception classes
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw ${amount}. Balance: ${balance}")

class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance

account = BankAccount(100)
try:
    account.withdraw(150)
except InsufficientFundsError as e:
    print(f"Error: {e}")
    print(f"  Shortfall: ${e.amount - e.balance}")
```

## 12.5 — Context Managers (`with` statement)
Guarantee cleanup even if errors occur.

```python
# File handling — the safe way
with open("test.txt", "w") as f:
    f.write("Hello, World!")
# File is AUTOMATICALLY closed here, even if write() raises an error

with open("test.txt", "r") as f:
    content = f.read()
    print(content)

# 🔥 Trick: Multiple context managers in one 'with'
with open("test.txt", "r") as source, open("copy.txt", "w") as dest:
    dest.write(source.read())
print("File copied!")
```

## 12.6 — 🔥 Writing Your Own Context Manager

```python
from contextlib import contextmanager
import time

@contextmanager
def timer(label="Block"):
    start = time.perf_counter()
    print(f"⏱️ {label} started...")
    yield  # Code inside 'with' runs here
    elapsed = time.perf_counter() - start
    print(f"⏱️ {label} finished in {elapsed:.4f}s")

# Use it!
with timer("Computation"):
    total = sum(range(1_000_000))
    print(f"  Sum = {total}")
```
