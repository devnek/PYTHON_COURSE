## 8.1 — Creating Sets

```python
# Sets — unordered, unique elements only
s1 = {1, 2, 3, 2, 1}     # Duplicates auto-removed
print(s1)                  # {1, 2, 3}

s2 = set([1, 2, 3, 4])    # From a list
s3 = set("mississippi")   # From a string → unique chars
print(s3)                  # {'m', 'i', 's', 'p'}

# ⚠️ Empty set — NOT {} (that's a dict!)
empty_set = set()
empty_dict = {}
print(type(empty_set))   # <class 'set'>
print(type(empty_dict))  # <class 'dict'>
```

## 8.2 — Set Operations
Sets support all mathematical set operations.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

# Union — elements in EITHER set
print(f"Union:              {a | b}")        # {1, 2, 3, 4, 5, 6, 7, 8}
print(f"Union (method):     {a.union(b)}")

# Intersection — elements in BOTH sets
print(f"Intersection:       {a & b}")        # {4, 5}
print(f"Intersection:       {a.intersection(b)}")

# Difference — elements in a but NOT in b
print(f"Difference (a-b):   {a - b}")        # {1, 2, 3}
print(f"Difference (b-a):   {b - a}")        # {6, 7, 8}

# Symmetric Difference — elements in EITHER but NOT BOTH
print(f"Symmetric Diff:     {a ^ b}")        # {1, 2, 3, 6, 7, 8}

# Subset and Superset
print(f"{{1,2}} ⊆ {{1,2,3}}? {  {1,2} <= {1,2,3}  }")    # True (subset)
print(f"{{1,2,3}} ⊇ {{1,2}}? {  {1,2,3} >= {1,2}  }")    # True (superset)
```

## 8.3 — 🔥 Practical Set Tricks

```python
# Trick 1: Remove duplicates from a list (fastest way)
nums = [1, 3, 2, 1, 4, 3, 5, 2]
unique = list(set(nums))  # Note: order may change!
print("Unique:", unique)

# Trick 2: Find common elements between lists
list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
common = list(set(list1) & set(list2))
print("Common:", common)

# Trick 3: Check if two lists have ANY overlap
has_overlap = bool(set(list1) & set(list2))
print("Has overlap:", has_overlap)

# Trick 4: Find missing and extra items
required = {"name", "email", "password"}
provided = {"name", "email", "age"}
missing = required - provided
extra = provided - required
print(f"Missing fields: {missing}")  # {'password'}
print(f"Extra fields: {extra}")      # {'age'}

# Trick 5: Set comprehension
squares = {x**2 for x in range(-5, 6)}  # Only unique squares
print("Unique squares:", sorted(squares))

# Trick 6: Remove duplicates preserving order (using set as tracker)
items = ["b", "a", "c", "a", "b", "d"]
seen = set()
ordered_unique = []
for item in items:
    if item not in seen:
        seen.add(item)
        ordered_unique.append(item)
print("Ordered unique:", ordered_unique)
```

## 8.4 — Frozen Sets
Immutable sets — can be used as dict keys or inside other sets.

```python
fs = frozenset([1, 2, 3])
print(fs)

# Can be used as a dict key
permissions = {
    frozenset({"read"}): "viewer",
    frozenset({"read", "write"}): "editor",
    frozenset({"read", "write", "delete"}): "admin",
}

user_perms = frozenset({"read", "write"})
print(f"Role: {permissions[user_perms]}")  # editor
```
